Observando los tipos de datos de la base CLIMA, hay un punto crítico: las columnas de los meses (ENE, MAR, ABR, ..., DIC), la columna ANUAL y la columna ALTITUD (m) están cargadas como tipo object (texto) en lugar de numéricas (float64 o int64).

Esto sucede frecuentemente porque algunos números en el archivo original traen comas en lugar de puntos decimales (ej. "1.530,5" o "2,450"), espacios, o guiones para indicar datos faltantes. Si no los convertimos a tipo numérico, no se podra calcular medias, desviaciones ni graficar los histogramas correctamente.

```python
# 1. Lista de columnas numéricas que deben ser float
cols_meses = ['ENE', 'FEB', 'MAR', 'ABR', 'MAY', 'JUN', 'JUL', 'AGO', 'SEP', 'OCT', 'NOV', 'DIC', 'ANUAL']
cols_numericas = ['ALTITUD (m)'] + cols_meses

# 2. Limpieza de caracteres y casteo a numérico
for col in cols_numericas:
    if col in df_clima.columns:
        # Convertir a texto, quitar comas de miles/decimales y reemplazar por punto si aplica
        df_clima[col] = (
            df_clima[col]
            .astype(str)
            .str.replace(',', '.', regex=False)
            .str.replace(' ', '', regex=False)
        )
        # Convertir a float; los valores no válidos (texto o símbolos) pasarán a ser NaN (nulos)
        df_clima[col] = pd.to_numeric(df_clima[col], errors='coerce')

# 3. Normalización de variables categóricas / texto
cols_texto = ['PARÁMETRO', 'CATEGORIA', 'ESTACIÓN', 'MUNICIPIO', 'DEPARTAMENTO']
for col in cols_texto:
    if col in df_clima.columns:
        df_clima[col] = df_clima[col].astype(str).str.strip().str.title()

# Mantener PERIODO como tipo categórico
if 'PERIODO' in df_clima.columns:
    df_clima['PERIODO'] = df_clima['PERIODO'].astype('category')

print("--- TIPOS DE DATOS CORREGIDOS ---")
print(df_clima.dtypes)
```

# Detección de Nulos

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("--- CLIMA: DETECCIÓN DE VALORES NULOS ---")

# 1. Conteo total y porcentaje de nulos por columna
total_nulos = df_clima.isnull().sum()
porcentaje_nulos = (df_clima.isnull().sum() / len(df_clima)) * 100

# Crear DataFrame resumido solo con columnas que tengan datos nulos (o todas)
df_nulos = pd.DataFrame({
    'Total Nulos': total_nulos,
    'Porcentaje (%)': porcentaje_nulos.round(2)
})

# Filtrar para mostrar la tabla ordenada por mayor cantidad de nulos
df_nulos_filtrado = df_nulos[df_nulos['Total Nulos'] > 0].sort_values(by='Total Nulos', ascending=False)

if len(df_nulos_filtrado) > 0:
    print(f"Se encontraron valores nulos en {len(df_nulos_filtrado)} columnas:\n")
    display(df_nulos_filtrado)

    # 2. Mapa de calor para visualizar la presencia de nulos
    plt.figure(figsize=(12, 5))
    sns.heatmap(df_clima.isnull(), cbar=False, cmap='viridis', yticklabels=False)
    plt.title('Mapa de Calor de Datos Nulos en CLIMA (Líneas amarillas = Datos faltantes)', fontsize=12, fontweight='bold')
    plt.xlabel('Columnas')
    plt.tight_layout()
    plt.show()
else:
    print("✅ ¡Perfecto! La base de datos CLIMA no tiene valores nulos en ninguna de sus columnas.")
```

Estrategia de Imputación y Limpieza para CLIMA
Para los meses (ENE a DIC): Al ser tan pocos faltantes (<0.06%), se imputan mediante la media/mediana agrupada por el mismo PARÁMETRO (ej. Precipitación o Temperatura) y DEPARTAMENTO (o MUNICIPIO).

Para la columna ANUAL: Se recalculan sus valores vacíos sumando o promediando directamente las columnas de los 12 meses (según si el parámetro es acumulado como precipitación o un promedio como temperatura).

```python
cols_meses = ['ENE', 'FEB', 'MAR', 'ABR', 'MAY', 'JUN', 'JUL', 'AGO', 'SEP', 'OCT', 'NOV', 'DIC']

# 1. Imputación de meses faltantes usando la mediana por PARÁMETRO y DEPARTAMENTO
for col in cols_meses:
    df_clima[col] = df_clima.groupby(['PARÁMETRO', 'DEPARTAMENTO'])[col].transform(
        lambda x: x.fillna(x.median())
    )
    # Si aún quedan nulos por departamentos pequeños, imputar con la mediana global del PARÁMETRO
    df_clima[col] = df_clima.groupby('PARÁMETRO')[col].transform(
        lambda x: x.fillna(x.median())
    )

# 2. Recalcular ANUAL para los registros donde es nulo
es_precipitacion = df_clima['PARÁMETRO'].astype(str).str.upper().str.contains('PRECIPITA')

suma_mensual = df_clima[cols_meses].sum(axis=1)
promedio_mensual = df_clima[cols_meses].mean(axis=1)

# Convertir el resultado de np.where a una Serie de Pandas con el mismo índice
valores_recalculados = pd.Series(
    np.where(es_precipitacion, suma_mensual, promedio_mensual),
    index=df_clima.index
)

# Imputar la columna ANUAL
df_clima['ANUAL'] = df_clima['ANUAL'].fillna(valores_recalculados)

print("--- VERIFICACIÓN POST-IMPUTACIÓN EN CLIMA ---")
print(f"Total nulos restantes en CLIMA: {df_clima.isnull().sum().sum()}")
```

# Avanzamos con la limpieza de la base de CRÉDITO

```python
import pandas as pd
import numpy as np

# 1. Carga de la base CREDITO
df_credito = pd.read_csv("CREDITO.csv")

print("--- BASE DE DATOS: CREDITO ---")
print(f"Dimensiones: {df_credito.shape[0]:,} filas y {df_credito.shape[1]} columnas\n")

print("Primeras 5 filas:")
display(df_credito.head())

print("Últimas 5 filas:")
display(df_credito.tail())

print("Tipos de datos iniciales:")
print(df_credito.dtypes)
```

Observando la estructura de la base CREDITO (1,807,886 filas y 27 columnas), tenemos varios puntos clave que requieren corrección de tipos antes del análisis estadístico:

Variables Monetarias como object: Valor Inversion, Colocacion y Vlr Inic Garantia vienen formateadas como texto con puntos de miles o comas.

Coordenadas Geográficas como object: LATITUD y LONGITUD están registradas como cadenas debido a separadores mixtos (puntos y comas decimales).

Variables Numéricas de Código / Identificador: ID Depto, Id Munic y ID Rubro deben ser normalizadas.

Nombres de Columnas y Textos: Fuente Colocacion, Tipo Productor, Linea de Credito, Linea de Produccion, Destino de Credito y Genero deben limpiarse de espacios adicionales.

Paso 2.2: Normalización de Textos y Casteo de Tipos en CREDITO

```python
import pandas as pd
import numpy as np

# 1. Copia limpia para evitar fragmentación
df_credito = df_credito.copy()

# 2. Limpieza de variables MONETARIAS y PORCENTAJES
cols_monetarias = ['Valor Inversion', 'Colocacion', 'Vlr Inic Garantia', '% FAG']
for col in cols_monetarias:
    if col in df_credito.columns:
        df_credito[col] = (
            df_credito[col]
            .astype(str)
            .str.replace('$', '', regex=False)
            .str.replace('.', '', regex=False)  # Quitar separador de miles
            .str.replace(',', '.', regex=False)  # Convertir coma decimal a punto
            .str.strip()
        )
        df_credito[col] = pd.to_numeric(df_credito[col], errors='coerce')

# 3. Limpieza de COORDENADAS GEOGRÁFICAS
cols_coords = ['LATITUD', 'LONGITUD']
for col in cols_coords:
    if col in df_credito.columns:
        df_credito[col] = (
            df_credito[col]
            .astype(str)
            .str.replace(',', '.', regex=False)  # Normalizar coma a punto decimal
            .str.strip()
        )
        df_credito[col] = pd.to_numeric(df_credito[col], errors='coerce')

# 4. Conversión de TEMPORALES Y CÓDIGOS a float64 (soporta nulos de forma nativa sin errores de casting)
cols_num_codigos = ['Año', 'Mes', 'ID Depto', 'Id Munic', 'DEPTCOL', 'MUNCOL', 'ID Rubro', 'CANTIDAD']
for col in cols_num_codigos:
    if col in df_credito.columns:
        df_credito[col] = pd.to_numeric(df_credito[col], errors='coerce')

# 5. Normalización de cadenas de TEXTO / CATEGÓRICAS
cols_texto = df_credito.select_dtypes(include=['object']).columns
for col in cols_texto:
    df_credito[col] = df_credito[col].astype(str).str.strip().str.upper()
    df_credito[col] = df_credito[col].replace(['NAN', 'NONE', 'N/A', ''], np.nan)

print("--- CREDITO: TIPOS DE DATOS CORREGIDOS CON ÉXITO ---")
print(df_credito.dtypes)
```

```python
import pandas as pd
import numpy as np

# 1. Cargar el dataset EVAS
df_evas = pd.read_csv("EVAS.csv")

print("--- BASE DE DATOS: EVAS ---")
print(f"Dimensiones: {df_evas.shape[0]:,} filas y {df_evas.shape[1]} columnas\n")

print("Primeras 5 filas:")
display(df_evas.head())

print("Tipos de datos iniciales:")
print(df_evas.dtypes)
```

Cargado el dataset EVAS ($166,732$ filas y $18$ columnas).A partir de la inspección inicial:Variables numéricas clave con formato incorrecto: Las columnas agrarias Área sembrada, Área cosechada, Producción y Rendimiento están leídas como object debido al uso de comas decimales (ej. 8,00 o 127,44).Códigos geográficos: Código Dane departamento, Código Dane municipio y Año vienen como int64.

Paso 3.1: Conversión de Tipos y Normalización de Texto en EVAS

```python
import pandas as pd
import numpy as np

# 1. Copia limpia de seguridad
df_evas = df_evas.copy()

# 2. Conversión de variables numéricas con comas decimales a float64
cols_agronomicas = ['Área sembrada', 'Área cosechada', 'Producción', 'Rendimiento']

for col in cols_agronomicas:
    if col in df_evas.columns:
        df_evas[col] = (
            df_evas[col]
            .astype(str)
            .str.replace('.', '', regex=False)   # Eliminar miles si aplican
            .str.replace(',', '.', regex=False)   # Convertir coma decimal a punto
            .str.strip()
        )
        df_evas[col] = pd.to_numeric(df_evas[col], errors='coerce')

# 3. Conversión de códigos a enteros/flotantes numéricos
cols_codigos = ['Código Dane departamento', 'Código Dane municipio', 'Año', 'Código del cultivo']
for col in cols_codigos:
    if col in df_evas.columns:
        df_evas[col] = pd.to_numeric(df_evas[col], errors='coerce')

# 4. Normalización de variables de texto (Strip + UPPER)
cols_texto_evas = df_evas.select_dtypes(include=['object', 'string']).columns

for col in cols_texto_evas:
    df_evas[col] = df_evas[col].astype(str).str.strip().str.upper()
    df_evas[col] = df_evas[col].replace(['NAN', 'NONE', 'N/A', ''], np.nan)

print("--- EVAS: TIPOS DE DATOS CORREGIDOS CON ÉXITO ---")
print(df_evas.dtypes)
```

Paso 3.2: Detección de Valores Nulos en EVAS


```python
print("--- EVAS: DETECCIÓN DE VALORES NULOS ---")

total_nulos_evas = df_evas.isnull().sum()
porcentaje_nulos_evas = (total_nulos_evas / len(df_evas)) * 100

df_nulos_evas = pd.DataFrame({
    'Total Nulos': total_nulos_evas,
    'Porcentaje (%)': porcentaje_nulos_evas.round(2)
}).sort_values(by='Total Nulos', ascending=False)

df_nulos_evas_filtrado = df_nulos_evas[df_nulos_evas['Total Nulos'] > 0]

print(f"Columnas con valores nulos ({len(df_nulos_evas_filtrado)} de {df_evas.shape[1]}):")
display(df_nulos_evas_filtrado)
```

Paso 3.3: Ingeniería de Características + EDA Completo de EVAS

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configuración gráfica
sns.set_theme(style="whitegrid")
plt.rcParams.update({'font.size': 10, 'figure.titlesize': 14})

# 1. INGENIERÍA DE CARACTERÍSTICAS
# Porcentaje de área no cosechada / perdida
df_evas['PERDIDA_AREA'] = np.where(
    df_evas['Área sembrada'] > 0,
    ((df_evas['Área sembrada'] - df_evas['Área cosechada']) / df_evas['Área sembrada']) * 100,
    0.0
)
# Ajustar posibles valores negativos por descuadres menores a 0
df_evas['PERDIDA_AREA'] = df_evas['PERDIDA_AREA'].clip(lower=0.0)

# 2. MEDIDAS DE TENDENCIA CENTRAL Y DISPERSIÓN
cols_num_evas = ['Área sembrada', 'Área cosechada', 'Producción', 'Rendimiento', 'PERDIDA_AREA']

stats_evas = pd.DataFrame(index=cols_num_evas)
stats_evas['Media'] = df_evas[cols_num_evas].mean()
stats_evas['Mediana'] = df_evas[cols_num_evas].median()
stats_evas['Moda'] = df_evas[cols_num_evas].mode().iloc[0]
stats_evas['Desv. Estándar'] = df_evas[cols_num_evas].std()
stats_evas['IQR'] = df_evas[cols_num_evas].quantile(0.75) - df_evas[cols_num_evas].quantile(0.25)
stats_evas['Mínimo'] = df_evas[cols_num_evas].min()
stats_evas['Máximo'] = df_evas[cols_num_evas].max()
stats_evas['Sesgo (Skew)'] = df_evas[cols_num_evas].skew()
stats_evas['Curtosis'] = df_evas[cols_num_evas].kurt()

print("--- MEDIDAS DE TENDENCIA CENTRAL Y DISPERSIÓN (EVAS) ---")
display(stats_evas.round(2))

# 3. DETECCIÓN DE OUTLIERS (MÉTODO IQR)
print("\n--- CONTEO Y PORCENTAJE DE OUTLIERS EN EVAS (MÉTODO IQR) ---")
outliers_evas = []

for col in ['Área sembrada', 'Producción', 'Rendimiento']:
    Q1 = df_evas[col].quantile(0.25)
    Q3 = df_evas[col].quantile(0.75)
    IQR = Q3 - Q1
    lim_inf = Q1 - 1.5 * IQR
    lim_sup = Q3 + 1.5 * IQR

    cant_out = ((df_evas[col] < lim_inf) | (df_evas[col] > lim_sup)).sum()
    pct_out = (cant_out / len(df_evas)) * 100
    outliers_evas.append({
        'Variable': col,
        'Límite Inferior': lim_inf,
        'Límite Superior': lim_sup,
        'Cant. Outliers': cant_out,
        '% Outliers': round(pct_out, 2)
    })

display(pd.DataFrame(outliers_evas))

# 4. ANÁLISIS GRAPHICO UNIVARIADO Y MULTIVARIADO
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# Distribución del Rendimiento
sns.histplot(df_evas['Rendimiento'], ax=axes[0,0], kde=True, color='forestgreen', log_scale=True)
axes[0,0].set_title('Distribución del Rendimiento (Escala Log)')

# Boxplot de Área Sembrada
sns.boxplot(x=df_evas['Área sembrada'], ax=axes[0,1], color='mediumseagreen')
axes[0,1].set_xscale('log')
axes[0,1].set_title('Boxplot / Outliers: Área Sembrada (Escala Log)')

# Top 10 Grupos de Cultivo por Producción Total
top_grupos = df_evas.groupby('Grupo cultivo')['Producción'].sum().nlargest(10).reset_index()
sns.barplot(data=top_grupos, y='Grupo cultivo', x='Producción', ax=axes[1,0], palette='viridis', hue='Grupo cultivo', legend=False)
axes[1,0].set_title('Top 10 Grupos de Cultivo por Producción Total')

# Matriz de Correlación
sns.heatmap(df_evas[cols_num_evas].corr(), annot=True, fmt=".2f", cmap="YlGnBu", ax=axes[1,1])
axes[1,1].set_title('Matriz de Correlación (EVAS)')

plt.tight_layout()
plt.show()

# 5. GUARDAR DATASET LIMPIO
df_evas.to_csv('EVAS_limpio.csv', index=False, encoding='utf-8-sig')
print("✅ Archivo 'EVAS_limpio.csv' guardado exitosamente.")
```

```python
import pandas as pd
import numpy as np

# Cargar dataset PRECIOS
df_precios = pd.read_csv("PRECIOS.csv")

print("--- BASE DE DATOS: PRECIOS ---")
print(f"Dimensiones: {df_precios.shape[0]:,} filas y {df_precios.shape[1]} columnas\n")

print("Primeras 5 filas:")
display(df_precios.head())

print("Tipos de datos iniciales:")
print(df_precios.dtypes)
```

Cargado el dataset PRECIOS (92 filas y 58 columnas).Inspección InicialEstructura: Se trata de una serie temporal de índices de precios de insumos agropecuarios a nivel nacional.Tipos de datos: La columna Fecha es de tipo object (formato MM/YYYY), mientras que las 57 columnas numéricas restantes son índices de tipo float64.

# Paso 5.1: Conversión de Fechas, Normalización y Verificación de Nulos

```python
import pandas as pd
import numpy as np

# 1. Copia de seguridad
df_precios = df_precios.copy()

# 2. Conversión de Fecha a datetime (formato MM/YYYY)
df_precios['Fecha'] = pd.to_datetime(df_precios['Fecha'], format='%m/%Y', errors='coerce')

# 3. Ordenar cronológicamente por Fecha
df_precios = df_precios.sort_values(by='Fecha').reset_index(drop=True)

# 4. Creación de variables temporales derivadas (Año, Mes)
df_precios['Año'] = df_precios['Fecha'].dt.year
df_precios['Mes'] = df_precios['Fecha'].dt.month

# 5. Detección de Valores Nulos
print("--- PRECIOS: DETECCIÓN DE VALORES NULOS ---")
total_nulos_p = df_precios.isnull().sum()
pct_nulos_p = (total_nulos_p / len(df_precios)) * 100

df_nulos_p = pd.DataFrame({
    'Total Nulos': total_nulos_p,
    'Porcentaje (%)': pct_nulos_p.round(2)
}).sort_values(by='Total Nulos', ascending=False)

df_nulos_p_filtrado = df_nulos_p[df_nulos_p['Total Nulos'] > 0]

if len(df_nulos_p_filtrado) > 0:
    print(f"Columnas con nulos ({len(df_nulos_p_filtrado)} de {df_precios.shape[1]}):")
    display(df_nulos_p_filtrado)
else:
    print("✅ No se detectaron valores nulos (0 vacíos).")

print(f"\nRango temporal cubierto: Desde {df_precios['Fecha'].min().strftime('%m/%Y')} hasta {df_precios['Fecha'].max().strftime('%m/%Y')}")
```

Se detectaron 25 valores nulos (27.17%) en la variable Total_Otros, mientras que las 59 columnas restantes están completamente limpias ($0$ vacíos).Dado que PRECIOS es una serie de tiempo ordenada cronológicamente, la mejor estrategia para imputar valores faltantes en series temporales es la interpolación lineal (o propagación hacia adelante/atrás ffill/bfill), lo que preserva la tendencia histórica de la serie.

Paso 5.2: Imputación, Medidas Descriptivas y Análisis de Series Temporales

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams.update({'font.size': 10, 'figure.titlesize': 14})

# 1. IMPUTACIÓN DE NULOS EN SERIE TEMPORAL
df_precios['Total_Otros'] = df_precios['Total_Otros'].interpolate(method='linear').bfill().ffill()
print(f"--- IMPUTACIÓN COMPLETADA: Nulos restantes en Total_Otros: {df_precios['Total_Otros'].isnull().sum()} ---")

# 2. MEDIDAS DE TENDENCIA CENTRAL Y DISPERSIÓN
cols_agregados = ['Indice_total', 'Total_Fertilizantes', 'Total_Plaguicidas',
                  'Total_Herbicidas', 'Total_Fungicidas', 'Total_Insecticidas', 'Total_Otros']

stats_precios = pd.DataFrame(index=cols_agregados)
stats_precios['Media'] = df_precios[cols_agregados].mean()
stats_precios['Mediana'] = df_precios[cols_agregados].median()
stats_precios['Moda'] = df_precios[cols_agregados].mode().iloc[0]
stats_precios['Desv. Estándar'] = df_precios[cols_agregados].std()
stats_precios['Mínimo'] = df_precios[cols_agregados].min()
stats_precios['Máximo'] = df_precios[cols_agregados].max()
stats_precios['Sesgo (Skew)'] = df_precios[cols_agregados].skew()

print("\n--- MEDIDAS DE TENDENCIA CENTRAL Y DISPERSIÓN (PRECIOS) ---")
display(stats_precios.round(2))

# 3. INGENIERÍA DE CARACTERÍSTICAS: Variación Mensual (%)
for col in cols_agregados:
    df_precios[f'VAR_PCT_{col}'] = df_precios[col].pct_change() * 100

# 4. VISUALIZACIÓN DE SERIES TEMPORALES
fig, axes = plt.subplots(2, 1, figsize=(15, 10), sharex=True)

# Gráfico 1: Evolución de Índices Principales
axes[0].plot(df_precios['Fecha'], df_precios['Indice_total'], label='Índice General Total', color='black', linewidth=2, linestyle='--')
axes[0].plot(df_precios['Fecha'], df_precios['Total_Fertilizantes'], label='Total Fertilizantes', color='darkgreen', linewidth=2)
axes[0].plot(df_precios['Fecha'], df_precios['Total_Plaguicidas'], label='Total Plaguicidas', color='darkred', linewidth=2)
axes[0].set_title('Evolución Histórica de Índices de Precios de Insumos Agropecuarios (12/2018 - 07/2026)')
axes[0].set_ylabel('Índice de Precios')
axes[0].legend(loc='upper left')

# Gráfico 2: Desagregación por Subgrupo de Plaguicidas
axes[1].plot(df_precios['Fecha'], df_precios['Total_Herbicidas'], label='Herbicidas', color='blue')
axes[1].plot(df_precios['Fecha'], df_precios['Total_Fungicidas'], label='Fungicidas', color='orange')
axes[1].plot(df_precios['Fecha'], df_precios['Total_Insecticidas'], label='Insecticidas', color='purple')
axes[1].set_title('Evolución por Subgrupo de Plaguicidas')
axes[1].set_xlabel('Fecha')
axes[1].set_ylabel('Índice de Precios')
axes[1].legend(loc='upper left')

plt.tight_layout()
plt.show()

# 5. GUARDAR DATASET LIMPIO Y ENRIQUECIDO
df_precios.to_csv('PRECIOS_limpio.csv', index=False, encoding='utf-8-sig')
print("✅ Archivo 'PRECIOS_limpio.csv' guardado exitosamente.")
```

Paso 6.1: Carga Correcta, Conversión a Numérico y Detección de Nulos

Reemplazar valores nulos implícitos ("ND", "NO INDICA", espacios, comas por puntos).

Castiagar las columnas químicas a tipo float64.

Reportar el porcentaje de nulos por variable.

```python
import pandas as pd
import numpy as np

# 1. Cargar reemplazando texto no detectable por NaN
df_suelos = pd.read_csv("SUELOS.csv", low_memory=False)

# 2. Normalización de cadenas inicial en variables cualitativas
cols_texto = ['Departamento', 'Municipio', 'Cultivo', 'Estado', 'Tiempo de establecimiento', 'Topografia', 'Drenaje', 'Riego']
for col in cols_texto:
    if col in df_suelos.columns:
        df_suelos[col] = df_suelos[col].astype(str).str.strip().str.upper()
        df_suelos[col] = df_suelos[col].replace(['NAN', 'NONE', 'N/A', 'NO INDICA', 'ND', ''], np.nan)

# 3. Conversión de columnas químicas/físicas a numéricas (float64)
cols_numericas = [
    'pH agua:suelo', 'Materia organica', 'Fosforo Bray II', 'Azufre Fosfato monocalcico',
    'Acidez Intercambiable', 'Aluminio intercambiable', 'Calcio intercambiable',
    'Magnesio intercambiable', 'Potasio intercambiable', 'Sodio intercambiable',
    'Capacidad de intercambio cationico', 'Conductividad electrica', 'Hierro disponible Olsen',
    'Cobre disponible', 'Manganeso disponible Olsen', 'Zinc disponible Olsen', 'Boro disponible',
    'Hierro disponible doble acido', 'Cobre disponible doble acido',
    'Manganeso disponible doble acido', 'Zinc disponible doble acido'
]

for col in cols_numericas:
    if col in df_suelos.columns:
        # Reemplazar comas decimales por puntos y forzar a numérico
        df_suelos[col] = df_suelos[col].astype(str).str.replace(',', '.', regex=False)
        df_suelos[col] = df_suelos[col].replace(['ND', 'NAN', 'NONE', 'N/A', 'NO INDICA', ''], np.nan)
        df_suelos[col] = pd.to_numeric(df_suelos[col], errors='coerce')

# 4. Diagnóstico de Nulos
print("--- SUELOS: DETECCIÓN DE VALORES NULOS Y TIPOS CORREGIDOS ---")
total_nulos_s = df_suelos.isnull().sum()
pct_nulos_s = (total_nulos_s / len(df_suelos)) * 100

df_nulos_s = pd.DataFrame({
    'Total Nulos': total_nulos_s,
    'Porcentaje (%)': pct_nulos_s.round(2),
    'Tipo de Dato': df_suelos.dtypes
}).sort_values(by='Total Nulos', ascending=False)

display(df_nulos_s)
```

# Paso 6.2: Normalización, Imputación e Ingeniería de Características en SUELOS
Para corregir los nombres de columnas con inconsistencias de formato/mayúsculas (como capacidad de intercambio cationico o Hierro disponible olsen), realizar la imputación edáfica e integrar los indicadores agronómicos

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams.update({'font.size': 10, 'figure.titlesize': 14})

# 1. NORMALIZACIÓN DE NOMENCLATURA EN COLUMNAS
df_suelos.columns = df_suelos.columns.str.strip().str.title()

# 2. CONVERSIÓN DE COLUMNAS RESTANTES A NUMÉRICO (Si quedaron como object)
cols_a_float = ['Fósforo Bray Ii', 'Capacidad De Intercambio Cationico', 'Hierro Disponible Olsen', 'Zinc Disponible Doble  Acido']
for col in cols_a_float:
    if col in df_suelos.columns:
        df_suelos[col] = df_suelos[col].astype(str).str.replace(',', '.', regex=False)
        df_suelos[col] = pd.to_numeric(df_suelos[col], errors='coerce')

# 3. IMPUTACIÓN ROBUSTA DE VARIABLES EDÁFICAS CLAVE
# Se imputa primero por la mediana del Departamento; si persiste nulo, por la mediana global
cols_clave_suelo = ['Ph Agua:Suelo', 'Materia Organica', 'Fósforo Bray Ii',
                    'Calcio Intercambiable', 'Magnesio Intercambiable', 'Potasio Intercambiable']

for col in cols_clave_suelo:
    if col in df_suelos.columns:
        df_suelos[col] = df_suelos.groupby('Departamento')[col].transform(lambda x: x.fillna(x.median()))
        df_suelos[col] = df_suelos[col].fillna(df_suelos[col].median())

# 4. INGENIERÍA DE CARACTERÍSTICAS EDÁFICAS
# a) Relación Ca/Mg (Balance catiónico esencial para asimilación de nutrientes)
if 'Calcio Intercambiable' in df_suelos.columns and 'Magnesio Intercambiable' in df_suelos.columns:
    df_suelos['RELACION_CA_MG'] = (df_suelos['Calcio Intercambiable'] / (df_suelos['Magnesio Intercambiable'] + 0.001)).round(2)

# b) Clasificación del Nivel de Acidez del Suelo
if 'Ph Agua:Suelo' in df_suelos.columns:
    condiciones_ph = [
        (df_suelos['Ph Agua:Suelo'] < 5.5),
        (df_suelos['Ph Agua:Suelo'] >= 5.5) & (df_suelos['Ph Agua:Suelo'] <= 6.5),
        (df_suelos['Ph Agua:Suelo'] > 6.5)
    ]
    etiquetas_ph = ['FUERTEMENTE ÁCIDO', 'ÓPTIMO / MODERADO', 'ALCALINO / NEUTRO']
    df_suelos['CATEGORIA_PH'] = np.select(condiciones_ph, etiquetas_ph, default='DESCONOCIDO')

# 5. MEDIDAS DE TENDENCIA CENTRAL Y DISPERSIÓN
stats_suelos = pd.DataFrame(index=cols_clave_suelo)
stats_suelos['Media'] = df_suelos[cols_clave_suelo].mean()
stats_suelos['Mediana'] = df_suelos[cols_clave_suelo].median()
stats_suelos['Moda'] = df_suelos[cols_clave_suelo].mode().iloc[0]
stats_suelos['Desv. Estándar'] = df_suelos[cols_clave_suelo].std()
stats_suelos['Mínimo'] = df_suelos[cols_clave_suelo].min()
stats_suelos['Máximo'] = df_suelos[cols_clave_suelo].max()
stats_suelos['Sesgo (Skew)'] = df_suelos[cols_clave_suelo].skew()

print("--- MEDIDAS DE TENDENCIA CENTRAL Y DISPERSIÓN (SUELOS) ---")
display(stats_suelos.round(2))

# 6. VISUALIZACIONES EDÁFICAS Y ANÁLISIS MULTIVARIADO
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Histogramas/KDE del pH del suelo
sns.histplot(df_suelos['Ph Agua:Suelo'].dropna(), kde=True, ax=axes[0], color='chocolate', bins=30)
axes[0].axvline(5.5, color='red', linestyle='--', linewidth=2, label='Límite Acidez Crítica (5.5)')
axes[0].set_title('Distribución del pH en Muestras de Suelo')
axes[0].set_xlabel('pH (Agua:Suelo)')
axes[0].legend()

# Conteo por Categoria de Acidez
sns.countplot(data=df_suelos, x='CATEGORIA_PH', ax=axes[1], palette='YlOrBr',
              order=['FUERTEMENTE ÁCIDO', 'ÓPTIMO / MODERADO', 'ALCALINO / NEUTRO'])
axes[1].set_title('Distribución Agronómica de Acidez del Suelo')
axes[1].set_xlabel('Categoría de pH')

plt.tight_layout()
plt.show()

# 7. GUARDAR DATASET LIMPIO Y FINAL DE LA TUBERÍA
df_suelos.to_csv('SUELOS_limpio.csv', index=False, encoding='utf-8-sig')
print("✅ Archivo 'SUELOS_limpio.csv' guardado exitosamente.")
```

Paso 2.1: Script de Consolidación y Merge Multidimensional

```python
import pandas as pd
import numpy as np

# 1. Diccionario Oficial DIVIPOLA Colombia (incluyendo claves numéricas y de texto)
mapa_divipola = {
    '1': 'AMAZONAS', '01': 'AMAZONAS', '5': 'ANTIOQUIA', '05': 'ANTIOQUIA',
    '8': 'ATLÁNTICO', '08': 'ATLÁNTICO', '11': 'BOGOTÁ, D.C.', '13': 'BOLÍVAR',
    '15': 'BOYACÁ', '17': 'CALDAS', '18': 'CAQUETÁ', '19': 'CAUCA',
    '20': 'CESAR', '23': 'CÓRDOBA', '25': 'CUNDINAMARCA', '27': 'CHOCÓ',
    '41': 'HUILA', '44': 'LA GUAJIRA', '47': 'MAGDALENA', '50': 'META',
    '52': 'NARIÑO', '54': 'NORTE DE SANTANDER', '63': 'QUINDÍO', '66': 'RISARALDA',
    '68': 'SANTANDER', '70': 'SUCRE', '73': 'TOLIMA', '76': 'VALLE DEL CAUCA',
    '81': 'ARAUCA', '85': 'CASANARE', '86': 'PUTUMAYO', '88': 'SAN ANDRÉS',
    '91': 'AMAZONAS', '94': 'GUAINÍA', '95': 'GUAVIARE', '97': 'VAUPÉS', '99': 'VICHADA'
}

# Cargar las 6 bases
clima = pd.read_csv("CLIMA_limpio.csv")
credito = pd.read_csv("CREDITO_limpio.csv")
evas = pd.read_csv("EVAS_limpio.csv")
fertilizantes = pd.read_csv("FERTILIZANTES_limpio.csv")
precios = pd.read_csv("PRECIOS_limpio.csv")
suelos = pd.read_csv("SUELOS_limpio.csv")

def limpiar_y_mapear_dep(df):
    col_dep = [c for c in df.columns if 'DEP' in c.upper()][0]
    # Limpiar formato flotante (ej: '5.0' -> '5')
    serie = df[col_dep].astype(str).str.replace(r'\.0$', '', regex=True).str.strip()
    # Mapear mediante el diccionario; si ya es texto (ej: "ANTIOQUIA"), se conserva
    df['DEPARTAMENTO_NORM'] = serie.map(mapa_divipola).fillna(serie).str.upper()
    return df

# Aplicar estandarización a las bases con componente geográfico
clima = limpiar_y_mapear_dep(clima)
credito = limpiar_y_mapear_dep(credito)
evas = limpiar_y_mapear_dep(evas)
suelos = limpiar_y_mapear_dep(suelos)

# 2. Agregaciones por Departamento Normalizado
suelos_agrup = suelos.groupby('DEPARTAMENTO_NORM').agg(
    PH_PROMEDIO=('Ph Agua:Suelo', 'mean'),
    PCT_SUELOS_ACIDOS=('Ph Agua:Suelo', lambda x: (x < 5.5).mean() * 100),
    MATERIA_ORGANICA_MEDIANA=('Materia Organica', 'median'),
    FOSFORO_MEDIANA=('Fósforo Bray Ii', 'median')
).reset_index()

cols_num_clima = [c for c in clima.select_dtypes(include=[np.number]).columns if c not in ['Año', 'Anio', 'Mes', 'Código', 'Codigo']]
clima_agrup = clima.groupby('DEPARTAMENTO_NORM')[cols_num_clima].mean().reset_index()

col_rend = [c for c in evas.columns if 'REND' in c.upper() or 'PROD' in c.upper()][0]
col_perd = [c for c in evas.columns if 'PERD' in c.upper() or 'AREA' in c.upper()][-1]
evas_agrup = evas.groupby('DEPARTAMENTO_NORM').agg(
    RENDIMIENTO_PROMEDIO=(col_rend, 'mean'),
    TOTAL_PERDIDA_AREA=(col_perd, 'sum')
).reset_index()

col_val_cred = [c for c in credito.columns if 'VALOR' in c.upper() or 'MONTO' in c.upper()][0]
credito_agrup = credito.groupby('DEPARTAMENTO_NORM').agg(
    TOTAL_VALOR_CREDITO=(col_val_cred, 'sum'),
    OPERACIONES_CREDITO=('DEPARTAMENTO_NORM', 'count')
).reset_index()

# 3. Unificación (Outer Merge para evaluar la coincidencia final)
df_maestro_limpio = pd.merge(suelos_agrup, clima_agrup, on='DEPARTAMENTO_NORM', how='outer')
df_maestro_limpio = pd.merge(df_maestro_limpio, evas_agrup, on='DEPARTAMENTO_NORM', how='outer')
df_maestro_limpio = pd.merge(df_maestro_limpio, credito_agrup, on='DEPARTAMENTO_NORM', how='outer')

# 4. Agregar macro-variables de PRECIOS y FERTILIZANTES
df_maestro_limpio['INDICE_PRECIO_FERTILIZANTES_MEDIO'] = precios['Total_Fertilizantes'].mean()
df_maestro_limpio['INDICE_PRECIO_HERBICIDAS_MEDIO'] = precios['Total_Herbicidas'].mean()
df_maestro_limpio['TOTAL_INSUMOS_CATALOGO_NACIONAL'] = len(fertilizantes)

print(f"✅ Tabla Maestra Unificada. Total de entidades territoriales consolidadas: {df_maestro_limpio.shape[0]}")
display(df_maestro_limpio.head(10))

# Guardar
df_maestro_limpio.to_csv("TABLA_MAESTRA_INTEGRADA.csv", index=False, encoding='utf-8-sig')
```

```python
import unicodedata

# Función para quitar tildes
def quitar_tildes(texto):
    if isinstance(texto, str):
        texto = unicodedata.normalize('NFD', texto)
        texto = ''.join(c for c in texto if unicodedata.category(c) != 'Mn')
    return texto

# Aplicar eliminación de tildes a la columna geográfica en la tabla maestra
df_maestro_limpio['DEPARTAMENTO_NORM'] = df_maestro_limpio['DEPARTAMENTO_NORM'].apply(quitar_tildes)

# Reagrupar para consolidar los duplicados por tilde (ej. ATLANTICO / ATLÁNTICO)
df_maestro_final = df_maestro_limpio.groupby('DEPARTAMENTO_NORM').mean(numeric_only=True).reset_index()

print(f"🎯 Tabla Maestra Final Pulida. Total entidades consolidadas: {df_maestro_final.shape[0]}")
df_maestro_final.to_csv("TABLA_MAESTRA_INTEGRADA.csv", index=False, encoding='utf-8-sig')
```

Paso 2.3: Segmentación Regional por Clústeres (K-Means)
Para categorizar los 34 departamentos en Perfiles Agrofinancieros y Edafoclimáticos, ejecutaremos un modelo no supervisado de K-Means. Esto permitirá segmentar el país en 3 o 4 grupos de riesgo/oportunidad.

```python
# 1. Instalación e importación de librerías necesarias
!pip install adjustText -q

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from adjustText import adjust_text

# 2. Cargar la Tabla Maestra Unificada
df_maestro = pd.read_csv("TABLA_MAESTRA_INTEGRADA.csv")

# 3. Seleccionar variables representativas para la segmentación
cols_cluster = [
    'PH_PROMEDIO', 'PCT_SUELOS_ACIDOS', 'FOSFORO_MEDIANA',
    'RENDIMIENTO_PROMEDIO', 'TOTAL_PERDIDA_AREA', 'TOTAL_VALOR_CREDITO'
]

# Filtrar departamentos sin nulos en las variables elegidas para el clustering
df_cluster = df_maestro.dropna(subset=cols_cluster).copy()

# 4. Escalamiento de Datos (Estandarización Z-score)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_cluster[cols_cluster])

# 5. Aplicación de K-Means (3 Clústeres Sugeridos)
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df_cluster['CLUSTER'] = kmeans.fit_predict(X_scaled)

# Mapear nombres descriptivos para los Clústeres
perfiles = {
    0: 'Perfil 1: Alta Vulnerabilidad Edáfica / Riego Agro',
    1: 'Perfil 2: Alta Productividad y Absorción Crediticia',
    2: 'Perfil 3: Balance Moderado / Desarrollo Intermedio'
}
df_cluster['NOMBRE_CLUSTER'] = df_cluster['CLUSTER'].map(perfiles)

# 6. Resumen de Medias por Clúster
resumen_clusters = df_cluster.groupby('NOMBRE_CLUSTER')[cols_cluster].mean()
print("--- CARACTERIZACIÓN DE CLÚSTERES AGROFINANCIEROS ---")
display(resumen_clusters.round(2))

# 7. Guardar la tabla maestra con la asignación de clústeres
df_cluster.to_csv("TABLA_MAESTRA_CLUSTERS.csv", index=False, encoding='utf-8-sig')
print("✅ Archivo 'TABLA_MAESTRA_CLUSTERS.csv' guardado exitosamente.")

# --- VISUALIZACIÓN OPTIMIZADA (SIN SOLAPAMIENTO DE ETIQUETAS) ---

# Crear columna de Crédito en Billones de Pesos para mejor lectura
df_cluster['CREDITO_BILLONES'] = df_cluster['TOTAL_VALOR_CREDITO'] / 1e12

plt.figure(figsize=(14, 9), dpi=100)
sns.set_theme(style="whitegrid")

# Gráfico de dispersión con puntos más visibles
sns.scatterplot(
    data=df_cluster,
    x='PCT_SUELOS_ACIDOS',
    y='CREDITO_BILLONES',
    hue='NOMBRE_CLUSTER',
    style='NOMBRE_CLUSTER',
    s=140,
    palette='Set1',
    alpha=0.85
)

# Escala Logarítmica en el eje Y para separar las observaciones apiladas en la base
plt.yscale('log')

# Algoritmo de repulsión de texto para separar etiquetas
texts = []
for i in range(df_cluster.shape[0]):
    nombre_dep = df_cluster['DEPARTAMENTO_NORM'].iloc[i]
    x_pos = df_cluster['PCT_SUELOS_ACIDOS'].iloc[i]
    y_pos = df_cluster['CREDITO_BILLONES'].iloc[i]

    if not np.isnan(x_pos) and not np.isnan(y_pos) and y_pos > 0:
        texts.append(
            plt.text(x_pos, y_pos, nombre_dep, fontsize=9, fontweight='bold', alpha=0.8)
        )

# Aplicar el ajuste de posiciones con líneas guía finas
adjust_text(
    texts,
    arrowprops=dict(arrowstyle='-', color='gray', lw=0.5, alpha=0.6),
    expand_points=(1.5, 1.5),
    expand_text=(1.2, 1.2)
)

plt.title('Segmentación Regional: % Suelos Ácidos vs. Crédito FINAGRO (Escala Logarítmica)', fontsize=14, pad=15)
plt.xlabel('% Suelos Ácidos (< 5.5 pH) por Departamento', fontsize=11)
plt.ylabel('Total Valor Crédito FINAGRO (Billones de COP - Escala Log)', fontsize=11)
plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left', borderaxespad=0, fontsize=10)
plt.tight_layout()
plt.show()
```
